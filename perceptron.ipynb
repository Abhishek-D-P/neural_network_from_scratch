```python
import numpy as np 
import pandas as pd 
from matplotlib import pyplot as plt
import seaborn as sns
import random
```

```python
data = pd.DataFrame({
    'u':[0,1,0,1],
    'v':[0,0,1,1],
    'y':[1,0,0,1]
})

data.head()
```

```python
sns.scatterplot(data=data,x='u',y='v',hue='y')
```

```python
class Perceptron:
    def __init__(self,n_iter=1000,lr=0.01):
        self.n_iter = n_iter
        self.lr = lr
        

    def fit(self,X,y):
        self.X = np.array(X)
        self.y = np.array(y)
        self.features = self.X.shape[1]
        self.weights = np.random.rand(self.features,)
        self.bias = np.random.rand()
        print("Initial weights")
        print(self.weights)
        print(self.bias)
        self.gradient_descent()
        print("Final weights")
        print(self.weights)
        print(self.bias)


    def activation(self,input):
        return np.array([1 if i>0 else 0 for i in input])

    def predict(self):
        return self.activation(np.matmul(self.X,self.weights) + self.bias)

    def error(self):
        return self.y-self.predict()
    
    def gradient_descent(self):
        for iter in range(self.n_iter):
            error = self.error()
            self.weights = self.weights + np.matmul(error,self.X)*(self.lr/self.X.shape[0])
            self.bias = self.bias + (self.lr*np.sum(error)/self.X.shape[0])

```

```python
X = data.iloc[:,:-1]
y = data.iloc[:,-1]
p = Perceptron()

p.fit(X,y)
p.predict()

sns.scatterplot(data=data,x='u',y='v',hue='y')
plt.plot(np.linspace(-2,2),[-(p.bias + p.weights[0]*xi)/p.weights[1] for xi in np.linspace(-2,2)])
```
